## 4.21 Membangun JavaScript Dashboard

```python
# =====================================================
# CELL 30 : BUILD DASHBOARD SCRIPT
# =====================================================

print_header("BUILD DASHBOARD SCRIPT")

dashboard_script = """
<script>

const dashboardData = __JSON__;

function showSample(sampleId){

    const sample = dashboardData[sampleId];

    if(!sample) return;

    document.getElementById("title").innerHTML =
        sample.metadata.sample_id;

    document.getElementById("metadata").innerHTML = `
        <table style="width:100%;border-collapse:collapse">

        <tr>
            <td><b>Group</b></td>
            <td>${sample.metadata.analysis_group}</td>
        </tr>

        <tr>
            <td><b>Detail</b></td>
            <td>${sample.metadata.detail_group}</td>
        </tr>

        <tr>
            <td><b>Actual</b></td>
            <td>${sample.metadata.actual_sentiment}</td>
        </tr>

        <tr>
            <td><b>Prediction</b></td>
            <td>${sample.metadata.predicted_sentiment}</td>
        </tr>

        <tr>
            <td><b>Confidence</b></td>
            <td>${(sample.metadata.confidence*100).toFixed(2)}%</td>
        </tr>

        <tr>
            <td><b>Text</b></td>
            <td>${sample.metadata.text}</td>
        </tr>

        </table>
    `;

    document.getElementById("shap_panel").innerHTML =
        `<img src="${sample.assets.shap_png}" style="width:100%">`;

    document.getElementById("lime_panel").innerHTML =
        `<iframe src="${sample.assets.lime_html}"
                 style="width:100%;height:650px;border:none;">
         </iframe>`;

}

</script>
"""

dashboard_script = dashboard_script.replace(
    "__JSON__",
    dashboard_json_text
)

print()

print_info(
    f"Panjang Script : {len(dashboard_script):,} karakter"
)

print()

print_success(
    "Dashboard Script berhasil dibuat."
)
```

## 4.22 Menambahkan Event Sidebar

```python
# =====================================================
# CELL 31 : ATTACH SIDEBAR EVENT
# =====================================================

print_header("ATTACH SIDEBAR EVENT")

sidebar_html_event = ""

for group, items in sidebar_structure.items():

    sidebar_html_event += f"""
    <div class="sidebar-group">
        <h3>{group}</h3>
    """

    for item in items:

        sidebar_html_event += f"""
        <button class="sample-btn"
                onclick="showSample('{item['sample_id']}')">
            {item['title']}
        </button>
        """

    sidebar_html_event += "</div>"

print()

print_info(
    "Event onclick berhasil dipasang."
)

print()

print_success(
    "Sidebar siap digunakan."
)
```
