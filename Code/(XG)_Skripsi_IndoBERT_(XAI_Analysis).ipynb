# BAB 2 Analisis High Confidence

Bab ini menyajikan analisis terhadap sampel pada kategori **High Confidence**, yaitu sampel yang diprediksi dengan tingkat keyakinan tinggi oleh model IndoBERT. Setiap kategori sentimen terdiri atas lima sampel yang diurutkan berdasarkan nilai confidence tertinggi. Analisis dilakukan dengan menampilkan ringkasan karakteristik SHAP, metadata prediksi, teks ulasan, serta observasi terhadap setiap kelompok sampel.

## 2.1 High Confidence Positif

```python
# =====================================================
# CELL 7 : LOAD HIGH CONFIDENCE POSITIVE
# =====================================================

print_header("LOAD HIGH CONFIDENCE POSITIVE")

high_positive = representative_df[
    (representative_df["analysis_group"] == "High Confidence") &
    (representative_df["detail_group"] == "Positif")
].reset_index(drop=True)

print()

print_info(
    f"Jumlah Sampel : {len(high_positive)}"
)

display(
    high_positive[
        [
            "sample_id",
            "actual_sentiment",
            "predicted_sentiment",
            "confidence"
        ]
    ]
)

print()

print_success(
    "High Confidence Positif berhasil dimuat."
)
```

```python
# =====================================================
# CELL 8 : BUILD HIGH CONFIDENCE POSITIVE
# =====================================================

print_header("BUILD HIGH CONFIDENCE POSITIVE")

hc_positive = representative_df[
    (representative_df["analysis_group"] == "High Confidence") &
    (representative_df["detail_group"] == "Positif")
].copy()

hc_positive = hc_positive.sort_values(
    by="confidence",
    ascending=False
).reset_index(drop=True)

print()

print_info(
    f"Jumlah Sampel : {len(hc_positive)}"
)

display(
    hc_positive[
        [
            "sample_id",
            "actual_sentiment",
            "predicted_sentiment",
            "confidence"
        ]
    ]
)

print()

print_success(
    "High Confidence Positif berhasil dipersiapkan."
)
```

### Ringkasan High Confidence Positif

```python
# =====================================================
# CELL 9 : SUMMARY HIGH CONFIDENCE POSITIVE
# =====================================================

print_header("SUMMARY HIGH CONFIDENCE POSITIVE")

summary_positive = hc_positive.merge(

    summary_df,

    on="sample_id",

    how="left",

    suffixes=("", "_shap")

)

summary_positive = summary_positive[

    [

        "sample_id",

        "confidence",

        "top_token",

        "top_positive_token",

        "top_negative_token",

        "mean_abs_shap",

        "max_abs_shap",

        "text"

    ]

]

display(summary_positive)

print()

print_success(

    "Ringkasan High Confidence Positif berhasil dibuat."

)
```

### Statistik High Confidence Positif

```python
# =====================================================
# CELL 10 : STATISTIC HIGH CONFIDENCE POSITIVE
# =====================================================

print_header("STATISTIC HIGH CONFIDENCE POSITIVE")

print()

print_info(
    f"Jumlah Sampel            : {len(summary_positive)}"
)

print_info(
    f"Confidence Maksimum      : {summary_positive['confidence'].max():.6f}"
)

print_info(
    f"Confidence Minimum       : {summary_positive['confidence'].min():.6f}"
)

print_info(
    f"Rata-rata Confidence     : {summary_positive['confidence'].mean():.6f}"
)

print_info(
    f"Rata-rata Mean |SHAP|    : {summary_positive['mean_abs_shap'].mean():.6f}"
)

print_info(
    f"Rata-rata Max |SHAP|     : {summary_positive['max_abs_shap'].mean():.6f}"
)

print()

top_token = (

    summary_positive["top_token"]

    .value_counts()

    .idxmax()

)

print_info(
    f"Top Token Dominan        : {top_token}"
)

print()

print_success(
    "Statistik berhasil dihitung."
)
```

### Lima Ulasan High Confidence Positif

```python
# =====================================================
# CELL 11 : HIGH CONFIDENCE POSITIVE DETAIL
# =====================================================

print_header("HIGH CONFIDENCE POSITIVE DETAIL")

hc_positive_detail = representative_df[
    (representative_df["analysis_group"] == "High Confidence") &
    (representative_df["detail_group"] == "Positif")
].copy()

hc_positive_detail = hc_positive_detail.sort_values(
    by="confidence",
    ascending=False
).reset_index(drop=True)

print()

print_info(
    f"Jumlah Sampel : {len(hc_positive_detail)}"
)

display(hc_positive_detail)

print()

print_success(
    "Detail High Confidence Positif berhasil dimuat."
)
```
